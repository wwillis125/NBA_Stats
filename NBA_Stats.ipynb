```python
def event_data(game):
    #get all gameplay eventst to see score changes
    gameplay = []
    for event in game['plays']:
        play_id = event['id']
        clock = event['clock']['displayValue']
        period = event['period']['number']
        awayscore = event['awayScore']
        homescore = event['homeScore']
        gameplay.append({'clock' : clock, 'period' : period, 'awayscore' : awayscore, 
                        'homescore': homescore, 'spread' : awayscore - homescore})
    #convert time on clock to seconds that have passed in the game
    for gametime in gameplay:
        #need to run an excception error because of how seconds are calculated in gameplay
        try:
            minutes, seconds = map(int, gametime['clock'].split(':'))
            total_minutes = (int(gametime['period']) - 1) * 720 + 720 -(minutes*60 + seconds)
            gametime['seconds'] = total_minutes

        except ValueError:
            gametime['seconds'] = (int(gametime['period']) - 1) * 720 + 720 - float(gametime['clock'])
    return pd.DataFrame(gameplay)
```
